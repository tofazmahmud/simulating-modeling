# Necessary Libraries

```python
import numpy as np
from sympy import Matrix
import matplotlib.pyplot as plt
import seaborn as sns
```

# Task 1 (Create my Personal Matrix)


```python
# d1(last digit of my student ID) = 9
# d2(second last digit of my student ID) = 4

A = np.array([[11, 5], [18, 6]])

```

# Task 2

```python
#1. shape of A
size_A = A.shape

# 2. Determinant of A
det_A = np.linalg.det(A)

# 3. Inverse of A
if np.linalg.det(A) != 0:
    inv_A = np.linalg.inv(A)
else:
    inv_A = "Matrix is singular, cannot compute inverse"

# 4. Rank of A
rank_A = np.linalg.matrix_rank(A)

# 5. Eigenvalues
eigenvalues, eigenvectors = np.linalg.eig(A)

#display results
print("Shape of A:", size_A)
print("Determinant of A:", det_A)
print("Inverse of A:\n", inv_A)
print("Rank of A:", rank_A)
print("Eigenvalues of A:", eigenvalues)
print("Eigenvectors of A:\n", eigenvectors)
```

# Task 3(Value change experiment)

```python
A = np.array([[11, 5], [18, 6]])

#Create a new matrix B
B = np.copy(A)
B[0][1] = B[0][1] + 1

print("B:\n", B)
```

#Task 4(Re-analyze Matrix B)

```python
#1. shape of B
size_B = B.shape

# 2. Determinant of B
det_B = np.linalg.det(B)

# 3. Inverse of B
if np.linalg.det(B) != 0:
    inv_B = np.linalg.inv(B)
else:
    inv_B = "Matrix is singular, cannot compute inverse"

# 4. Rank of B
rank_B = np.linalg.matrix_rank(A)

# 5. Eigenvalues
eigenvalues, eigenvectors = np.linalg.eig(B)

#display results
print("Shape of B:", size_B)
print("Determinant of B:", det_B)
print("Inverse of B:\n", inv_B)
print("Rank of B:", rank_B)
print("Eigenvalues of B:", eigenvalues)
print("Eigenvectors of B:\n", eigenvectors)
```
