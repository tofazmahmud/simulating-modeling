# Chi Square Test Question Solution
Chi-Square Test Solution in Python

## Import Necessary Libraries

```python
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
```

##1: My Student Id's last two digit 49

##2: Observed Data

```python
d = 49

a = 20 + d         # C
b = 25 + (d % 10)  # C++
c = 30 + (d % 7)   # Java
e = 35 + (d % 5)   # Python
observed = np.array([a,b,c,e])


print("My last two digit of id: ", d)
print("O : ", observed)


```

## 3,4: Null & Alternative Hypothesis
H0 : Students have equal preferences for all four programming languages (25% each).

H1 : Students do not have equal preferences across all four languages.

##5: Expected frequencies for each category

```python
total = np.sum(observed)
expected = np.full(len(observed), total / len(observed), dtype=float)
print(f'Total: {total}')
print(f'Expected frequency per language: {total / len(observed):.2f}')
print(f'Expected frequencies: {expected}')
```

##6: Chi Square static calculation

```python
chi_square_stat = np.sum(((observed - expected) ** 2) / expected)
```

##7: Degrees of freedom

```python
df = len(observed) - 1
```

## 8: Calculate the p-value and critical value

```python
#p_value = 1 - stats.chi2.cdf(chi_square_stat, df)  # p-value
alpha = 0.05  # Significance level
critical_value = stats.chi2.ppf(1 - alpha, df)  # Critical value
```

##9: Final Decision

```python
print(f'Chi-square statistic: {chi_square_stat:.2f}')
print(f'Degrees of freedom: {df}')
print(f'Critical value (0.05 significance level): {critical_value:.2f}')
#print(f'P-value: {p_value:.4f}')
```

## Conclusion

```python
if chi_square_stat > critical_value:
    print('Reject the null hypothesis. Observed data does not match expected distribution.')
else:
    print('Fail to reject the null hypothesis. Observed data matches the expected distribution.')
```

## Plotting observed vs expected frequencies

```python
plt.bar(range(1, len(observed) + 1), observed, width=0.4, label='Observed', alpha=0.7)
plt.bar(np.arange(1, len(observed) + 1) + 0.4, expected, width=0.4, label='Expected', alpha=0.7) #You can’t do range(...) + 0.4 because range objects don’t support adding a float to every element directly.
plt.xlabel('Categories')
plt.ylabel('Frequencies')
plt.title('Observed vs Expected Frequencies')
plt.legend()
plt.show()
```
